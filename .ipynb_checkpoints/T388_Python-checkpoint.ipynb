```python
# BREAK
for i in range(1,101):
    if i==12:
        break
    else:
        print(i)
print("program ends !")        
```

```python
# CONTINOUS
#Write a progam (1 to 10) but skip 7 if encountred 
i=1
while i<10:
    if i==7:
        i=i+1
        continue

    else:
        print(i)
        i+=1
```

```python
#write to print the string entered by user .BUt it must skip all the VOWELS in it
character = input("ENter the word of your choice: ")
# VOWELS: a,e,i,o,u
for i in character:

    if i =="a" or i=="e" or i=="i" or i=="o" or i=="u":
        continue

    else:
        print(i)
```

```python
# if we have to give the value which is capital like "S" in continous write this code(in continous it will not take continous value)
character = input("ENter the word of your choice: ")

print(character)
for i in character:
    j = i.lower()
    if j=="a" or j=="e" or j=="i" or j=="o" or j=="u":
        continue

    else:
        print(i) 
```

# write a program to find temperature in celcius from the farenheit input 

$C = (F-32)5/9$

```python
def celcius(Farenheit):
    c = (Farenheit-32) *(5/9)
    print(f"For farenheit{Farenheit}, Temperature in celcius is {c}")  #here we also use "F String"

celcius(98)    
celcius(102)
```

```python
# ROUND FUNCTION 
def celcius(Farenheit):
    c = (Farenheit-32) *(5/9)
    print(f"For farenheit{Farenheit}, Temperature in celcius is {round(c,2)}") #here we are using round fun..by using this..in output where 36.66666 it will give 36.67 the exact output
    #here we also use "F String"
    
celcius(98)    
celcius(102)
```

# write to print whether a 3 digit number is a Armstrong Number 

```python
num = int(input("Enter any number to check whether Armstrong or not: "))

length = len(str(num))
result = 0
for i in str(num):
    result= result + (int(i)**length)

if result == num:
    print("this number",num,"is an Armstrong Number!")
else:
    print("oh!,this number is not an Armstrong Number")
    
```

# FACTORIAL

```python
# 5! =5x4x3x2x1 = 120
# 3! =3x2x1 = 6
def fact(n):
    if n==1:
        return 1
    else:
        product = n*fact(n-1)
        return product
fact(4)        
```

```python
                               #replace KIWI with pineapple using for i in range ,if else


fruits= ["Apple","apple","Kiwi","Cherry","tamato","Kiwi"]
#index      0       1       2       3        4       5
#element    1       2       3       4        5       6

for i in range(0,6):
    if fruits[i]=="Kiwi":
        fruits[i] = "Pineapple"

fruits        
```

```python
# Here we are creating new_list which is blank and then we are doing append(add value) one by one in the list

fruits= ["Apple","apple","Kiwi","Cherry","tamato","Kiwi"]
#index      0       1       2       3        4       5
#element    1       2       3       4        5       6
New_list = []

for i in fruits:
    if i == "Kiwi":
        i="Pineapple"
        New_list.append(i)
    else:
          New_list.append(i)

New_list    
```

```python
# here we are using ELIF....as we are using if to append kiwi with pineapple and we have to also append Apple with 
# gauav then we have to use ELIF

fruits= ["Apple","apple","Kiwi","Cherry","tamato","Kiwi"]
#index      0       1       2       3        4       5
#element    1       2       3       4        5       6
New_list = []

for i in fruits:
    if i == "Kiwi":
        i="Pineapple"
        New_list.append(i)
    elif i== "Apple":
        i="Guava"
        New_list.append(i)
    else:
          New_list.append(i)

New_list  
```
